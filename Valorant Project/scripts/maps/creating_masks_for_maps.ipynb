```python
### Extracting Maps From Images

import cv2
import os
import numpy as np 


INPUT_DIR = r"C:\Users\scott\Documents\Project-To-Do\Valorant Project\maps\raw"
OUTPUT_DIR = r"C:\Users\scott\Documents\Project-To-Do\Valorant Project\maps\extracted"
os.makedirs(OUTPUT_DIR, exist_ok=True)


SIZE = 384
RADIUS = SIZE // 2
X_OFFSET = 15  # pixels from left
Y_OFFSET = 35  # pixels from top

# Create circular mask once
circle_mask = np.zeros((SIZE, SIZE), dtype=np.uint8)
cv2.circle(circle_mask, (RADIUS, RADIUS), RADIUS, 255, -1)

for fname in os.listdir(INPUT_DIR):
    if not fname.lower().endswith((".png", ".jpg", ".jpeg")):
        continue

    img = cv2.imread(os.path.join(INPUT_DIR, fname), cv2.IMREAD_COLOR)
    
    # Correct offset crop
    crop = img[Y_OFFSET:Y_OFFSET + SIZE, X_OFFSET:X_OFFSET + SIZE]

    # Convert to BGRA to add transparency
    crop = cv2.cvtColor(crop, cv2.COLOR_BGR2BGRA)
    crop[:, :, 3] = circle_mask  # alpha channel

    out_path = os.path.join(OUTPUT_DIR, os.path.splitext(fname)[0] + ".png")
    cv2.imwrite(out_path, crop)
```

```python
### Creating the Walkable Areas vs Non Walkable Areas for Mask


import cv2
import numpy as np

# Load your minimap

INPUT_DIR = r"C:\Users\scott\Documents\Project-To-Do\Valorant Project\maps\extracted"
OUTPUT_DIR = r"C:\Users\scott\Documents\Project-To-Do\Valorant Project\maps\walkable_mask"


# Convert to grayscale
for fname in os.listdir(INPUT_DIR):
    if not fname.lower().endswith((".png", ".jpg", ".jpeg")):
        print(f"Not Useable {fname}")
        continue
    
    print(fname)
    img = cv2.imread(os.path.join(INPUT_DIR, fname), cv2.IMREAD_COLOR)
    
    #Converting to Gray 
    
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    
    _, walkable_mask = cv2.threshold(gray, 100, 255, cv2.THRESH_BINARY)

    # Optional: mask outside circle (radius=192, center=(192,192))
    h, w = gray.shape
    circle_mask = np.zeros_like(walkable_mask)
    cv2.circle(circle_mask, (w//2, h//2), w//2, 255, -1)
    walkable_mask = cv2.bitwise_and(walkable_mask, circle_mask)
    
    # Cleaning Up Output
            
    # kernel = np.ones((3, 3), np.uint8)  # small kernel
    # walkable_clean = cv2.morphologyEx(walkable_mask, cv2.MORPH_OPEN, kernel)
    # walkable_clean = cv2.morphologyEx(walkable_clean, cv2.MORPH_CLOSE, kernel)

    # walkable_clean = cv2.bitwise_and(walkable_clean, circle_mask)

    # walkable_clean = cv2.GaussianBlur(walkable_clean, (3, 3), 0)
    # _, walkable_clean = cv2.threshold(walkable_clean, 127, 255, cv2.THRESH_BINARY)

    out_path = os.path.join(OUTPUT_DIR, os.path.splitext(fname)[0] + ".png")
    
    cv2.imwrite(out_path, walkable_mask)
```
